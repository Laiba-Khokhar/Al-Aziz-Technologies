```python
import numpy as np
# Arrays
prices=np.array([5000,10000,15000,20000])
new_prices = prices * 1.10
print(new_prices)

```

```python
# Vectorization
prices = np.array([500, 1000, 1500, 2000, 2500])
quantity = np.array([2, 3, 4, 2, 5])

total_sales = prices * quantity

print("Total Sales:", total_sales)

total_revenue = np.sum(total_sales)

print("Total Revenue:", total_revenue)
```

```python
# Arrays operations

sales = np.array([10000, 20000, 30000, 40000, 50000])
print("Total:", np.sum(sales))
print("Average:", np.mean(sales))
print("Maximum:", np.max(sales))
print("Minimum:", np.min(sales))

increased_sales = sales * 1.10
print("Increased Sales:", increased_sales)
```

```python
# Indexing

sales = np.array([10000, 20000, 30000, 40000, 50000])
print(sales[0])
print(sales[-1])
print(sales[2])
print(sales[1:4])
```

```python
# Pandas Series
import pandas as pd
prices = pd.Series(
    [500, 1000, 1500, 2000, 2500],
    name="Price"
)
print(prices)

print("Total:", prices.sum())
print("Average:", prices.mean())
print("Maximum:", prices.max())
print("Minimum:", prices.min())
```

```python
# creating CSV File
import pandas as pd

data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Electronics"],
    "Price": [85000, 1500, 3000, 25000, 18000],
    "Quantity": [2, 5, 3, 2, 1]
}

df = pd.DataFrame(data)

df.to_csv("products.csv", index=False)
df = pd.read_csv("products.csv")

df
pd.read_csv("products.csv")
```

```python
# Reading the excel files
df.to_excel("products.xlsx", index=False)
df_excel = pd.read_excel("products.xlsx")

df_excel
print(df_excel.shape)
```

```python
# Merging Datasets
customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4],
    "Customer_Name": ["Ali", "Sara", "Ahmed", "Ayesha"]
})
sales = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 1],
    "Sales": [50000, 30000, 45000, 20000]
})
merged_df = pd.merge(customers, sales, on="Customer_ID")

merged_df
```

```python

# Concatenation

sales_part1 = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard"],
    "Sales": [85000, 1500, 3000]
})

sales_part2 = pd.DataFrame({
    "Product": ["Monitor", "Printer"],
    "Sales": [25000, 18000]
})
combined_sales = pd.concat(
    [sales_part1, sales_part2],
    ignore_index=True
)

combined_sales
```
