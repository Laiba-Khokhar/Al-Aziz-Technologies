```python
import pandas as pd

sales = pd.Series([10000, 12000, 15000, 11000, 13000, 14000, 90000])

Q1 = sales.quantile(0.25)
Q3 = sales.quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = sales[(sales < lower_limit) | (sales > upper_limit)]

print("Outliers:")
print(outliers)
```

```python
data = {
    "Customer": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Sales": [5000, 15000, 8000, 25000]
}

df_apply = pd.DataFrame(data)

df_apply
```

```python
def sales_level(sales):
    if sales >= 10000:
        return "High"
    else:
        return "Low"
```

```python
df_apply["Sales_Level"] = df_apply["Sales"].apply(sales_level)

df_apply
```

```python
import pandas as pd

data = {
    "Customer": ["  Ali  ", "Sara", "Ali", "AHMED", "Ayesha"],
    "Age": ["25", "30", "25", None, "28"],
    "City": ["Lahore", "Karachi", "Lahore", " lahore ", None],
    "Sales": [5000, 15000, 5000, 8000, None]
}

df_messy = pd.DataFrame(data)

print("Original Data:")
display(df_messy)

# 1. Remove duplicate records
df_messy = df_messy.drop_duplicates()

# 2. Fix Age data type
df_messy["Age"] = pd.to_numeric(df_messy["Age"], errors="coerce")

# 3. Fill missing Age with the average
df_messy["Age"] = df_messy["Age"].fillna(df_messy["Age"].mean())

# 4. Fill missing City
df_messy["City"] = df_messy["City"].fillna("Unknown")

# 5. Clean text
df_messy["Customer"] = df_messy["Customer"].str.strip().str.title()
df_messy["City"] = df_messy["City"].str.strip().str.title()

# 6. Fill missing Sales
df_messy["Sales"] = df_messy["Sales"].fillna(df_messy["Sales"].mean())

print("\nCleaned Data:")
display(df_messy)
```
